```kotlin
enum class Semaphore(val text: String) {
    RED("Stop"), YELLOW("Caution"), GREEN("Go")
}
val state: Semaphore = Semaphore.YELLOW
println(state)
println(state.ordinal)
println(state.name)
println(Semaphore.entries)
for(e in Semaphore.entries)
    println("${e.name} - ${e.ordinal} -> ${e.text}")
```
